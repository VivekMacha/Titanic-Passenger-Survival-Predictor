```python
import pandas as pd
from tensorflow.keras.models import load_model
import pickle
import tensorflow as tf
```

```python
with open('label_encoder.pkl','rb') as file:
    label = pickle.load(file)

with open('onehot_encoder.pkl','rb') as file:
    onehot = pickle.load(file)

with open('scaler.pkl','rb') as file:
    scaler = pickle.load(file)
```

```python
model = load_model('model.h5')
```

```python
user = {'Pclass':3,'Sex':'male','SibSp':1,'Parch':0,'Fare':76,'Embarked':'Southampton'}
user = pd.DataFrame([user])
```

```python
user
```

```python
user['Sex'] = label.transform(user['Sex'])
```

```python
embarked = onehot.transform(user[['Embarked']])
```

```python
embarked = pd.DataFrame(embarked,columns=onehot.get_feature_names_out())
```

```python
embarked
```

```python
df = pd.concat([user.drop(columns=['Embarked']),embarked],axis=1)
```

```python
df
```

```python
df[['Pclass','SibSp','Parch','Fare']] = scaler.transform(df[['Pclass','SibSp','Parch','Fare']])
```

```python
df
```

```python
y = model.predict(df)
```
