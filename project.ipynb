## Data Cleaning

```python
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
df['Embarked'].replace({'C':'Cherbourg','S':'Southampton','Q':'Queenstown'},inplace=True)
```

# Data Encoding & Scaling
- Label Encoder
  - Binary
- Ordinal Encoder
  - Ordinal Data
- Nominal Encoder
  - Normal data

```python
from sklearn.preprocessing import LabelEncoder,OneHotEncoder
```

```python
labelencoder = LabelEncoder()
onehotencoder = OneHotEncoder(sparse_output=False)
```

```python
df['Sex'] = labelencoder.fit_transform(df['Sex'])
```

```python
embarked = onehotencoder.fit_transform(df[['Embarked']])
```

```python
embarked
```

```python
embarked = pd.DataFrame(embarked,columns=onehotencoder.get_feature_names_out())
```

```python
df = pd.concat([df,embarked],axis=1)
```

```python
df.drop(columns=['Embarked'],inplace=True)
```

```python
df
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
ss = StandardScaler()
```

```python
num_cols = ['Pclass','SibSp','Parch','Fare']
```

```python
df[num_cols] = ss.fit_transform(df[num_cols])
```

## Train-Validation-Test Split

```python
X = df.drop(columns=['Survived'])
Y = df['Survived']
```

```python
X
```

```python
Y
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
x_train,x_valid,y_train,y_valid =train_test_split(x_train,y_train,test_size=0.2,random_state=42)
```

```python
x_train
```

```python
import tensorflow
```

# Model Building

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping
```

```python
input_shape = x_train.shape[1]
```

```python
model = Sequential(
    [
        Dense(64,input_shape=(input_shape,),activation='relu'),
        Dense(32,activation='relu'),
        Dense(1,activation='sigmoid')
    ]
)
```

```python
model.summary()
```

```python
opt = tensorflow.keras.optimizers.Adam(learning_rate= 0.01)
```

```python
model.compile(optimizer=opt,loss='binary_crossentropy',metrics=['accuracy'])
```

```python
stopping_callback = EarlyStopping(
    monitor='val_loss',
    patience=5,
    restore_best_weights=True
)
```

```python
model.fit(x_train,y_train,validation_data=(x_valid,y_valid),epochs=10,callbacks=stopping_callback)
```

```python
model.save('model.h5')
```
